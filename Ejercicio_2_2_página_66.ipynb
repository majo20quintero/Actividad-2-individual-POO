```python

from enum import Enum


# Enumeración para el tipo de planeta
class TipoPlaneta(Enum):
    GASEOSO = "GASEOSO"
    TERRESTRE = "TERRESTRE"
    ENANO = "ENANO"


class Planeta:
    # Constante: distancia de una unidad astronómica en kilómetros
    UA_KM = 149597870

    # Constructor
    def __init__(
        self,
        nombre=None,
        cantidad_satélites=0,
        masa=0.0,
        volumen=0.0,
        diámetro=0,
        distancia_media_sol=0,
        tipo_planeta=TipoPlaneta.TERRESTRE,
        observable_simple_vista=False
    ):
        self.nombre = nombre
        self.cantidad_satélites = cantidad_satélites
        self.masa = masa
        self.volumen = volumen
        self.diámetro = diámetro
        self.distancia_media_sol = distancia_media_sol
        self.tipo_planeta = tipo_planeta
        self.observable_simple_vista = observable_simple_vista

    # Método para imprimir los atributos
    def imprimir(self):
        print("Nombre:", self.nombre)
        print("Cantidad de satélites:", self.cantidad_satélites)
        print("Masa (kg):", self.masa)
        print("Volumen (km³):", self.volumen)
        print("Diámetro (km):", self.diámetro)
        print(
            "Distancia media al Sol (millones de km):",
            self.distancia_media_sol
        )
        print("Tipo de planeta:", self.tipo_planeta.value)
        print(
            "Observable a simple vista:",
            "Sí" if self.observable_simple_vista else "No"
        )

    # Método para calcular la densidad
    def calcular_densidad(self):
        if self.volumen == 0:
            return None

        # Convertir el volumen de km³ a m³ para expresar
        # la densidad en kg/m³
        volumen_m3 = self.volumen * 1_000_000_000
        return self.masa / volumen_m3

    # Método para determinar si el planeta es exterior
    def es_planeta_exterior(self):
        # La distancia está expresada en millones de kilómetros.
        # El límite exterior comienza después de 3.4 UA.
        limite_exterior = 3.4 * self.UA_KM / 1_000_000

        return self.distancia_media_sol > limite_exterior


# Método principal
def main():
    # Creación del primer planeta: Tierra
    tierra = Planeta(
        nombre="Tierra",
        cantidad_satélites=1,
        masa=5.972e24,
        volumen=1.08321e12,
        diámetro=12742,
        distancia_media_sol=150,
        tipo_planeta=TipoPlaneta.TERRESTRE,
        observable_simple_vista=True
    )

    # Creación del segundo planeta: Júpiter
    jupiter = Planeta(
        nombre="Júpiter",
        cantidad_satélites=95,
        masa=1.898e27,
        volumen=1.4313e15,
        diámetro=139820,
        distancia_media_sol=778,
        tipo_planeta=TipoPlaneta.GASEOSO,
        observable_simple_vista=True
    )

    planetas = [tierra, jupiter]

    for planeta in planetas:
        print("\n========== DATOS DEL PLANETA ==========")
        planeta.imprimir()

        densidad = planeta.calcular_densidad()

        if densidad is None:
            print("Densidad: no se puede calcular porque el volumen es cero")
        else:
            print(f"Densidad: {densidad:.2f} kg/m³")

        if planeta.es_planeta_exterior():
            print("¿Es un planeta exterior?: Sí")
        else:
            print("¿Es un planeta exterior?: No")


if __name__ == "__main__":
    main()
```

```python

from enum import Enum


# Enumeración del tipo de planeta
class TipoPlaneta(Enum):
    GASEOSO = "GASEOSO"
    TERRESTRE = "TERRESTRE"
    ENANO = "ENANO"


class Planeta:
    # Constante: una unidad astronómica en kilómetros
    UA_KM = 149597870

    # Constructor
    def __init__(
        self,
        nombre=None,
        cantidad_satélites=0,
        masa=0.0,
        volumen=0.0,
        diámetro=0,
        distancia_media_sol=0,
        tipo_planeta=TipoPlaneta.TERRESTRE,
        observable_simple_vista=False,
        periodo_orbital=0.0,
        periodo_rotacion=0.0
    ):
        self.nombre = nombre
        self.cantidad_satélites = cantidad_satélites
        self.masa = masa
        self.volumen = volumen
        self.diámetro = diámetro
        self.distancia_media_sol = distancia_media_sol
        self.tipo_planeta = tipo_planeta
        self.observable_simple_vista = observable_simple_vista

        # Nuevos atributos
        self.periodo_orbital = periodo_orbital
        self.periodo_rotacion = periodo_rotacion

    # Método para imprimir los atributos
    def imprimir(self):
        print("Nombre:", self.nombre)
        print("Cantidad de satélites:", self.cantidad_satélites)
        print("Masa (kg):", self.masa)
        print("Volumen (km³):", self.volumen)
        print("Diámetro (km):", self.diámetro)
        print(
            "Distancia media al Sol (millones de km):",
            self.distancia_media_sol
        )
        print("Tipo de planeta:", self.tipo_planeta.value)
        print(
            "Observable a simple vista:",
            "Sí" if self.observable_simple_vista else "No"
        )

        # Mostrar los nuevos atributos
        print("Período orbital (años):", self.periodo_orbital)
        print("Período de rotación (días):", self.periodo_rotacion)

    # Método para calcular la densidad en kg/m³
    def calcular_densidad(self):
        if self.volumen == 0:
            return None

        volumen_m3 = self.volumen * 1_000_000_000
        return self.masa / volumen_m3

    # Método para determinar si es un planeta exterior
    def es_planeta_exterior(self):
        limite_exterior = 3.4 * self.UA_KM / 1_000_000
        return self.distancia_media_sol > limite_exterior


# Método principal
def main():
    # Primer planeta: Tierra
    tierra = Planeta(
        nombre="Tierra",
        cantidad_satélites=1,
        masa=5.972e24,
        volumen=1.08321e12,
        diámetro=12742,
        distancia_media_sol=150,
        tipo_planeta=TipoPlaneta.TERRESTRE,
        observable_simple_vista=True,
        periodo_orbital=1.0,
        periodo_rotacion=0.9973
    )

    # Segundo planeta: Júpiter
    jupiter = Planeta(
        nombre="Júpiter",
        cantidad_satélites=95,
        masa=1.898e27,
        volumen=1.4313e15,
        diámetro=139820,
        distancia_media_sol=778,
        tipo_planeta=TipoPlaneta.GASEOSO,
        observable_simple_vista=True,
        periodo_orbital=11.86,
        periodo_rotacion=0.4135
    )

    planetas = [tierra, jupiter]

    for planeta in planetas:
        print("\n========== DATOS DEL PLANETA ==========")
        planeta.imprimir()

        densidad = planeta.calcular_densidad()

        if densidad is None:
            print("Densidad: no se puede calcular porque el volumen es cero")
        else:
            print(f"Densidad: {densidad:.2f} kg/m³")

        if planeta.es_planeta_exterior():
            print("¿Es un planeta exterior?: Sí")
        else:
            print("¿Es un planeta exterior?: No")


if __name__ == "__main__":
    main()
```
