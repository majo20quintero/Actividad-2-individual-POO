```python

class Persona:
    # Constructor de la clase
    def __init__(self, nombre, apellido, documento, año_nacimiento):
        self.nombre = nombre
        self.apellido = apellido
        self.documento = documento
        self.año_nacimiento = año_nacimiento

    # Método para imprimir los atributos de la persona
    def imprimir_datos(self):
        print("Nombre:", self.nombre)
        print("Apellido:", self.apellido)
        print("Documento de identidad:", self.documento)
        print("Año de nacimiento:", self.año_nacimiento)
        print("-----------------------------")


# Método principal: creación de dos personas
def main():
    persona1 = Persona("María", "López", "123456789", 2004)
    persona2 = Persona("Carlos", "Pérez", "987654321", 2002)

    print("DATOS DE LA PERSONA 1")
    persona1.imprimir_datos()

    print("DATOS DE LA PERSONA 2")
    persona2.imprimir_datos()


# Ejecutar el programa
if __name__ == "__main__":
    main()
```

```python

class Persona:
    # Constructor de la clase
    def __init__(self, nombre, apellido, documento,
                 anio_nacimiento, pais_nacimiento, genero):
        self.nombre = nombre
        self.apellido = apellido
        self.documento = documento
        self.anio_nacimiento = anio_nacimiento
        self.pais_nacimiento = pais_nacimiento
        self.genero = genero

    # Método para imprimir los atributos
    def imprimir(self):
        print("Nombre:", self.nombre)
        print("Apellido:", self.apellido)
        print("Documento de identidad:", self.documento)
        print("Año de nacimiento:", self.anio_nacimiento)
        print("País de nacimiento:", self.pais_nacimiento)
        print("Género:", self.genero)
        print("-----------------------------")


# Método principal
def main():
    persona1 = Persona(
        "María", "López", "123456789",
        2004, "Colombia", "M"
    )

    persona2 = Persona(
        "Carlos", "Pérez", "987654321",
        2002, "Colombia", "H"
    )

    print("DATOS DE LA PERSONA 1")
    persona1.imprimir()

    print("DATOS DE LA PERSONA 2")
    persona2.imprimir()


if __name__ == "__main__":
    main()
```
